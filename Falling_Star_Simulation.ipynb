```python
import numpy as np
import matplotlib.pyplot as plt

# Falling Star Velocity Simulation using Euler's method
def falling_star_velocity_optimized(mass, height, time_step=0.1, air_resistance=0.02):
    g = 9.81 # gravity in m/s^2
    velocity = 0 # initial velocity
    position = height # initial height
    times = [0] # time list for plotting
    velocities = [velocity] # velocity list for plotting
    positions = [position] # position list for plotting

    while position > 0:
        # velocity with air resistance and gravity
        acceleration = g - air_resistance * velocity # includes resistance
        velocity += acceleration * time_step
        position -= velocity * time_step


        # Store data
        times.append(times[-1] + time_step)
        velocities.append(velocity)
        positions.append(position)

    return times, positions, velocities

# Example simulation
times, positions, velocities = falling_star_velocity_optimized(mass=1, height=1000)

# Plotting the optimized velocity and position
plt.figure(figsize=(12, 6))

# Velocity plot
plt.subplot(2, 1, 1)
plt.plot(times, positions, label='Position (m)',color= 'r')
plt.title('Position of Falling Star')
plt.xlabel('Time (seconds)')
plt.ylabel('Position (m)')
plt.grid(True)

plt.tight_layout()
plt.show()
```
